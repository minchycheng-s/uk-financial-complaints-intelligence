# 01 — FCA workbook profiling review

This notebook reviews generated profiling outputs only. All profiling logic lives in `src/customer_harm/profiling`; raw workbooks are never written. Run `profile-fca-workbooks` before opening this notebook.

```python
from pathlib import Path
import json
import pandas as pd

PROFILE_DIR = Path('../data/interim/profiling')
workbooks = pd.read_csv(PROFILE_DIR / 'workbook_inventory.csv')
sheets = pd.read_csv(PROFILE_DIR / 'sheet_inventory.csv')
columns = pd.read_csv(PROFILE_DIR / 'column_profiles.csv')
headers = pd.read_csv(PROFILE_DIR / 'header_detection_results.csv')
schema = pd.read_csv(PROFILE_DIR / 'schema_comparison.csv')
warnings = pd.read_csv(PROFILE_DIR / 'profiling_warnings.csv')
formats = pd.read_csv(PROFILE_DIR / 'cell_format_profiles.csv')
product_groups = pd.read_csv(PROFILE_DIR / 'product_group_inventory.csv')
product_group_changes = pd.read_csv(PROFILE_DIR / 'product_group_comparison.csv')
taxonomy = pd.read_csv(PROFILE_DIR / 'product_taxonomy_inventory.csv')
taxonomy_changes = pd.read_csv(PROFILE_DIR / 'product_taxonomy_comparison.csv')
definitions = pd.read_csv(PROFILE_DIR / 'reporting_definitions.csv')
definition_changes = pd.read_csv(PROFILE_DIR / 'reporting_definition_comparison.csv')
review_register = pd.read_csv(PROFILE_DIR / 'profiling_review_register.csv')
review_gate = json.loads((PROFILE_DIR / 'profiling_review_gate.json').read_text())
```

## Profiling review gate

```python
display(pd.DataFrame([review_gate]))
display(review_register[['review_item_id', 'category', 'scope', 'severity', 'blocks_extraction', 'decision_status', 'recommended_action']])
```

## Product definitions and reporting wording

```python
display(product_groups.groupby(['reporting_period', 'label_type']).size().unstack(fill_value=0))
display(product_group_changes)
display(taxonomy.groupby(['reporting_period', 'inherited_product_group']).size().unstack(fill_value=0))
display(taxonomy_changes)
display(definition_changes[['previous_period', 'current_period', 'definition_key', 'change_type']])
```

## Source coverage and processing status

```python
display(workbooks[['reporting_period', 'source_file_name', 'sheet_count', 'rows_extracted', 'processing_status', 'warning_count']])
display(sheets.groupby('reporting_period').agg(sheets=('sheet', 'size'), rows=('data_rows', 'sum'), columns=('data_columns', 'sum')))
```

## Header detection requiring review

```python
selected = headers.query('selected == True')
display(selected.sort_values('score').head(20)[['reporting_period', 'sheet', 'row_number', 'score', 'values']])
display(sheets.query('duplicate_column_count > 0 or empty_leading_rows > 0')[['reporting_period', 'sheet', 'header_row', 'duplicate_columns', 'empty_leading_rows']])
```

## Missingness and workbook formatting

```python
display(columns.query('row_count > 0').nlargest(25, 'missing_percent')[['reporting_period', 'sheet', 'column', 'semantic_type', 'null_count', 'blank_string_count', 'missing_percent']])
display(columns.groupby(['reporting_period', 'semantic_type']).size().unstack(fill_value=0))
marker_columns = ['missing_marker_count', 'suppressed_marker_count', 'not_applicable_marker_count', 'excel_error_count']
display(columns.loc[columns[marker_columns].sum(axis=1) > 0, ['reporting_period', 'sheet', 'column', 'semantic_type'] + marker_columns])
display(formats.groupby(['reporting_period', 'number_format_class'])['formatted_cell_count'].sum().unstack(fill_value=0))
```

## Cross-period schema changes and warnings

```python
display(schema.groupby(['current_period', 'change_type']).size().unstack(fill_value=0))
display(schema)
display(warnings)
```
